Step 2: Missing values analysis

```python
df.isnull().sum()
```

Step 3: Duplicate-value analysis

```python
df.duplicated().sum()
```

```python
df['Order_Date']= pd.to_datetime(df['Order_Date'])
df.dtypes
```

```python
print('Start date:',df['Order_Date'].min())
print('End date:',df['Order_Date'].max())
```

Step 4: Key Performance Indicator.

```python
print('Total Sales:',df['Total_Sales'].sum())
print('Total Profit: ', df['Profit'].sum())
print('Total Quantity Sold:', df['Quantity'].sum())
print('Total Orders:', df['Order_ID'].nunique())
print('Total Customers:', df['Customer_Name'].nunique())

```

```python
#average_order_value= df['Total_Sales'].mean()
#rint('Average Order Value:', average_order_value)
```

```python
average_order_value=df['Total_Sales'].sum()/df['Order_ID'].nunique()
print('Average Order Value:', average_order_value)
```

**SALES BY  PRODUCT CATEGORY**

```python
category_sales=df.groupby('Product_Category')['Total_Sales'].sum()
print('Category Sales:')
print(category_sales)
```

**PROFIT BY PRODUCT CATEGORY**

```python
category_profit=df.groupby('Product_Category')['Profit'].sum()
print('Category Profit:')
print(category_profit)
```

**SALES BY REGION**

```python
region_sales=df.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
print('Region Sales:')
print(region_sales)
```

**PROFIT BY REGION**

```python
region_profit=df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
print('Region Profit:')
print(region_profit)
```

**Top 10 Products By Sales**

```python
product_sales=df.groupby('Product_Name')['Total_Sales'].sum().sort_values(ascending=False).head(10)
print('Top 10 Products by Sales:')
print(product_sales)
```

**Customer Segment analysis**

```python
segment_sales=df.groupby('Customer_Segment')['Total_Sales'].sum().sort_values(ascending=False)
print('Customer Segment Sales:')
print(segment_sales)
```

```python
segment_profit=df.groupby('Customer_Segment')['Profit'].sum().sort_values(ascending=False)
print('Customer Segment Profit:')
print(segment_profit)
```

**Monthly Sales Analysis**

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'])
df['Month']=df['Order_Date'].dt.month
monthly_sales=df.groupby('Month')['Total_Sales'].sum()
print('Monthly Sales:')
print(monthly_sales)
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

monthly_profit = df.groupby(df['Order_Date'].dt.month)['Profit'].sum()

print('Monthly Profit:')
print(monthly_profit)
```

**Payment Method analysis**

```python
payment_sales = df.groupby('Payment_Method')['Total_Sales'].sum()

print('Sales by Payment Method:')
print(payment_sales)
```

```python
payment_profit = df.groupby('Payment_Method')['Profit'].sum()

print('Profit by Payment Method:')
print(payment_profit)
```

Q**uantity Sold by Product Category**

```python
category_quantity = df.groupby('Product_Category')['Quantity'].sum()

print('Quantity Sold by Category:')
print(category_quantity)
```

**Sales by Country**

```python
country_sales = df.groupby('Country')['Total_Sales'].sum().sort_values(ascending=False)

print('Sales by Country:')
print(country_sales)
```

**Profit by Country**

```python
country_profit = df.groupby('Country')['Profit'].sum().sort_values(ascending=False)

print("Profit by Country:")
print(country_profit)
```
